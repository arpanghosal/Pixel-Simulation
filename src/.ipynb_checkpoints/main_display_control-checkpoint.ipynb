```python
# Ensure we can import from our 'src' package
import sys
import os

# Add the parent directory (project root) to the system path
# This allows Jupyter to find the 'src' package
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

# Now import our simulated device
from src.simulated_display import SimulatedLEDDisplay

import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create an instance of our simulated display
# Let's make it a 16x16 display
display_width = 16
display_height = 16
my_display_module = SimulatedLEDDisplay(display_width, display_height, initial_color=(50, 0, 50))

# Read some "firmware" info
print(f"Firmware Version: {my_display_module.get_firmware_version()}")
```

```python
# Clear the display first
my_display_module.clear_display()
```

```python
# Draw a red square in the middle
red = (255, 0, 0)
for y in range(4, 12):
    for x in range(4, 12):
        my_display_module.set_pixel(x, y, red)

# Get the current display data
display_image_data = my_display_module.get_display_data()

# Visualize the display data
plt.figure(figsize=(4, 4))
plt.imshow(display_image_data)
plt.title("Simulated LED Display")
plt.axis('off') # Hide axes
plt.show()
```

```python
# Clear again and draw a blue diagonal line
my_display_module.fill_display((0, 0, 50)) # Dark blue background
blue = (0, 0, 255)
for i in range(min(display_width, display_height)):
    my_display_module.set_pixel(i, i, blue)
    my_display_module.set_pixel(display_width - 1 - i, i, (255, 255, 0)) # Yellow anti-diagonal

display_image_data = my_display_module.get_display_data()
plt.figure(figsize=(4, 4))
plt.imshow(display_image_data)
plt.title("Simulated LED Display - Diagonals")
plt.axis('off')
plt.show()

# Simulate a hardware reset
my_display_module.reset_hardware()
display_image_data = my_display_module.get_display_data()
plt.figure(figsize=(4, 4))
plt.imshow(display_image_data)
plt.title("Simulated LED Display - After Reset")
plt.axis('off')
plt.show()
```
